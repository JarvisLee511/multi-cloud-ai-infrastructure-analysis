```python
import pandas as pd
import plotly.express as px
```

```python
market_share = pd.DataFrame({
    "year_quarter": [
        "2024 Q4",
        "2025 Q1",
        "2025 Q2",
        "2025 Q3",
        "2025 Q4",
        "2026 Q1"
    ],
    "AWS": [
        30,
        32,
        30,
        29,
        28,
        28
    ],
    "Azure": [
        21,
        21,
        20,
        20,
        21,
        21
    ],
    "GCP": [
        11,
        12,
        13,
        13,
        15,
        15
    ]
})

market_share
```

```python
market_share_long = market_share.melt(
    id_vars="year_quarter",
    var_name="provider",
    value_name="market_share"
)

market_share_long
```

```python
fig = px.line(
    market_share_long,
    x="year_quarter",
    y="market_share",
    color="provider",
    markers=True,
    title="Cloud Infrastructure Market Share Trend"
)

fig.update_layout(
    yaxis_title="Market Share (%)",
    xaxis_title="Year / Quarter"
)

fig.show()
```

```python
market_share.to_csv(
    "cloud_market_share_wide.csv",
    index=False
)

market_share_long.to_csv(
    "cloud_market_share_long.csv",
    index=False
)
```
