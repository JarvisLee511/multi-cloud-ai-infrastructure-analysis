```python
import pandas as pd
from pathlib import Path

CURATED_DIR = Path("curated_data")

service_catalog = pd.read_csv(
    CURATED_DIR / "native_cloud_service_catalog_v3.csv"
)

service_summary_v3 = (
    service_catalog
    .groupby("provider")
    .agg(service_count=("service_name", "nunique"))
    .reset_index()
)

service_category_summary_v3 = (
    service_catalog
    .groupby(["provider", "category"])
    .agg(service_count=("service_name", "nunique"))
    .reset_index()
)

service_summary_v3.to_csv(
    CURATED_DIR / "service_summary_v3.csv",
    index=False
)

service_category_summary_v3.to_csv(
    CURATED_DIR / "service_category_summary_v3.csv",
    index=False
)

print(service_summary_v3)
print()
print(service_category_summary_v3.head(20))
```

```python
import pandas as pd
from pathlib import Path
import os

TABLEAU_DIR = Path("tableau_export")
TABLEAU_DIR.mkdir(exist_ok=True)

CLEANED_DIR = Path("cleaned_data")
CURATED_DIR = Path("curated_data")

market_share = pd.read_csv(CLEANED_DIR / "market_share_clean.csv")
region_master = pd.read_csv(CLEANED_DIR / "master_region_table.csv")
region_summary = pd.read_csv(CLEANED_DIR / "region_summary_clean.csv")
gpu_pricing = pd.read_csv(CLEANED_DIR / "gpu_pricing_summary_clean.csv")
aws_gpu = pd.read_csv(CLEANED_DIR / "aws_gpu_summary_clean.csv")

ai_taxonomy = pd.read_csv(CURATED_DIR / "ai_taxonomy_summary_v2.csv")
ai_services = pd.read_csv(CURATED_DIR / "ai_services_v2.csv")
service_catalog = pd.read_csv(CURATED_DIR / "native_cloud_service_catalog_v3.csv")
service_summary = pd.read_csv(CURATED_DIR / "service_summary_v3.csv")
service_category = pd.read_csv(CURATED_DIR / "service_category_summary_v3.csv")

market_share.to_csv(TABLEAU_DIR / "market_share.csv", index=False)
region_master.to_csv(TABLEAU_DIR / "region_master.csv", index=False)
region_summary.to_csv(TABLEAU_DIR / "region_summary.csv", index=False)
gpu_pricing.to_csv(TABLEAU_DIR / "gpu_pricing.csv", index=False)
aws_gpu.to_csv(TABLEAU_DIR / "aws_gpu_summary.csv", index=False)

ai_taxonomy.to_csv(TABLEAU_DIR / "ai_taxonomy.csv", index=False)
ai_services.to_csv(TABLEAU_DIR / "ai_services.csv", index=False)
service_catalog.to_csv(TABLEAU_DIR / "service_catalog.csv", index=False)
service_summary.to_csv(TABLEAU_DIR / "service_summary.csv", index=False)
service_category.to_csv(TABLEAU_DIR / "service_category_summary.csv", index=False)

print(os.listdir(TABLEAU_DIR))
```
