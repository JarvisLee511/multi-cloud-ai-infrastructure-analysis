```python
import re
import pandas as pd

native_service_catalog = pd.read_csv("native_cloud_service_catalog.csv")

ai_patterns = [
    r"\bgpu\b",
    r"\bai\b",
    r"\bml\b",
    r"machine learning",
    r"deep learning",
    r"foundation model",
    r"\bllm\b",
    r"training",
    r"inference",
    r"tensor",
    r"cuda",
    r"neural",
    r"computer vision",
    r"\bnlp\b",
    r"generative",
    r"genai",
    r"bedrock",
    r"vertex ai",
    r"openai",
    r"sagemaker",
    r"rekognition",
    r"comprehend",
    r"lookoutvision",
    r"hugging face",
    r"tpu"
]

exclude_patterns = [
    r"blockchain",
    r"mail",
    r"trail",
    r"lightsail",
    r"supplychain",
    r"container instances",
    r"active directory",
    r"domain controller",
    r"gateway",
    r"redis",
    r"rails"
]

ai_regex = "|".join(ai_patterns)
exclude_regex = "|".join(exclude_patterns)

service_text = native_service_catalog["service_name"].str.lower()

ai_df = native_service_catalog[
    service_text.str.contains(ai_regex, regex=True, na=False) &
    ~service_text.str.contains(exclude_regex, regex=True, na=False)
].copy()

ai_df[["provider", "service_name", "category"]].head(50)
```

```python
provider_ai_count = (
    ai_df.groupby("provider")
    .size()
    .reset_index(name="ai_service_count")
    .sort_values("ai_service_count", ascending=False)
)

provider_ai_count
```

```python
total_services = (
    native_service_catalog
    .groupby("provider")
    .size()
    .reset_index(name="total_native_services")
)

provider_ai_share = provider_ai_count.merge(
    total_services,
    on="provider"
)

provider_ai_share["ai_service_share_pct"] = (
    provider_ai_share["ai_service_count"] /
    provider_ai_share["total_native_services"] * 100
)

provider_ai_share
```

```python
ai_df.to_csv("ai_gpu_related_services.csv", index=False)
provider_ai_share.to_csv("provider_ai_service_share.csv", index=False)
```

```python
def classify_service_origin(service_name):
    text = str(service_name).lower()

    native_keywords = [
        "amazon", "aws", "azure", "microsoft",
        "google", "cloud", "vertex", "gemini",
        "bedrock", "sagemaker", "openai"
    ]

    partner_keywords = [
        "byol", "bitnami", "weka", "h2o.ai",
        "thoughtspot", "labs.ai", "axi ai",
        "barracuda", "nvidia image", "jupyter",
        "pytorch", "tensorflow"
    ]

    if any(k in text for k in partner_keywords):
        return "Marketplace / Partner"
    elif any(k in text for k in native_keywords):
        return "Native Cloud Service"
    else:
        return "Unclear"

ai_df["service_origin"] = ai_df["service_name"].apply(classify_service_origin)

ai_origin_summary = (
    ai_df.groupby(["provider", "service_origin"])
    .size()
    .reset_index(name="service_count")
)

ai_origin_summary
```

```python
native_ai_df = ai_df[
    ai_df["service_origin"] == "Native Cloud Service"
].copy()
```

```python
native_ai_count = (
    native_ai_df.groupby("provider")
    .size()
    .reset_index(name="native_ai_service_count")
)

native_ai_count
```

```python
ai_df.to_csv("ai_gpu_related_services.csv", index=False)
provider_ai_share.to_csv("provider_ai_service_share.csv", index=False)
```

```python
def classify_ai_type(service_name):

    text = str(service_name).lower()

    # Foundation Models
    if any(k in text for k in [
        "bedrock",
        "openai",
        "gemini",
        "llm",
        "foundation",
        "generative"
    ]):
        return "Foundation Models"

    # ML Platform
    elif any(k in text for k in [
        "sagemaker",
        "vertex ai",
        "machine learning",
        "training"
    ]):
        return "ML Platform"

    # GPU / Infrastructure
    elif any(k in text for k in [
        "gpu",
        "tpu",
        "cuda",
        "inferentia",
        "trainium"
    ]):
        return "AI Infrastructure"

    # Vision
    elif any(k in text for k in [
        "vision",
        "rekognition",
        "image"
    ]):
        return "Computer Vision"

    # NLP
    elif any(k in text for k in [
        "nlp",
        "language",
        "translate",
        "speech",
        "comprehend"
    ]):
        return "Language AI"

    else:
        return "General AI"
```

```python
native_ai_df["ai_type"] = (
    native_ai_df["service_name"]
    .apply(classify_ai_type)
)
```

```python
ai_type_summary = (
    native_ai_df.groupby(
        ["provider", "ai_type"]
    )
    .size()
    .reset_index(name="service_count")
)

ai_type_summary
```

```python
native_ai_df.to_csv(
    "native_ai_services_enriched.csv",
    index=False
)

ai_type_summary.to_csv(
    "ai_type_summary.csv",
    index=False
)
```
