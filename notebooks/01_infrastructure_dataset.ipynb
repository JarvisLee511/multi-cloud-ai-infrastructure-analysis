```python
azure_regions = [
    ["Azure", "East US", "eastus", "North America"],
    ["Azure", "East US 2", "eastus2", "North America"],
    ["Azure", "Central US", "centralus", "North America"],
    ["Azure", "West US", "westus", "North America"],
    ["Azure", "West US 2", "westus2", "North America"],

    ["Azure", "Canada Central", "canadacentral", "North America"],
    ["Azure", "Brazil South", "brazilsouth", "South America"],

    ["Azure", "UK South", "uksouth", "Europe"],
    ["Azure", "France Central", "francecentral", "Europe"],
    ["Azure", "Germany West Central", "germanywestcentral", "Europe"],

    ["Azure", "Japan East", "japaneast", "Asia Pacific"],
    ["Azure", "Korea Central", "koreacentral", "Asia Pacific"],
    ["Azure", "Southeast Asia", "southeastasia", "Asia Pacific"],
    ["Azure", "Australia East", "australiaeast", "Oceania"],

    ["Azure", "UAE North", "uaenorth", "Middle East"],
    ["Azure", "South Africa North", "southafricanorth", "Africa"]
]
```

```python
azure_df = pd.DataFrame(
    azure_regions,
    columns=[
        "provider",
        "region_name",
        "region_code",
        "continent"
    ]
)

azure_df.head()
```

```python
azure_df["is_public"] = True
```

```python
azure_df.info()
```

```python
multi_cloud_df = pd.concat([
    aws_df,
    azure_df,
    gcp_df
], ignore_index=True)
print(multi_cloud_df)
```

```python
# Fix GCP region_code from region_name
gcp_df["region_code"] = gcp_df["region_name"].str.extract(r"^([a-z0-9-]+)")
gcp_df["service"] = "Cloud Platform"
gcp_df["is_public"] = True
gcp_df["status"] = "Available"

# Fix Azure columns
azure_df["service"] = "Cloud Platform"
azure_df["status"] = "Available"

# Fix AWS columns
aws_df["region_name"] = aws_df["region_code"]
aws_df["service"] = "EC2"
aws_df["is_public"] = True
aws_df["status"] = "Available"

def map_continent(code):
    if code.startswith(("us-", "ca-", "mx-")):
        return "North America"
    elif code.startswith("sa-"):
        return "South America"
    elif code.startswith("eu-"):
        return "Europe"
    elif code.startswith(("ap-",)):
        return "Asia Pacific"
    elif code.startswith("me-") or code.startswith("il-"):
        return "Middle East"
    elif code.startswith("af-"):
        return "Africa"
    else:
        return "Other"

aws_df["continent"] = aws_df["region_code"].apply(map_continent)

# Standardize final columns
final_columns = [
    "provider",
    "service",
    "region_code",
    "region_name",
    "continent",
    "is_public",
    "status"
]

multi_cloud_df = pd.concat([
    aws_df[final_columns],
    azure_df[final_columns],
    gcp_df[final_columns]
], ignore_index=True)

multi_cloud_df
```
